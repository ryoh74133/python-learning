```python
try:
  x=int(input("数字を入れて"))
  print(10/x)
except ValueError:
  print("数字じゃない")
except ZeroDivisionError:
  print("0はだめ")
```

```python
try:
  food=["ハンバーガー","ポテト","ナゲット"]
  n=int(input("何番目を表示する？"))
  print(food[n])
except ValueError:
  print("数字を入れて")
except IndexError:
  print("0から2までの数字を入れて")
```

```python
while True:
  try:
    age = int(input("年齢を入れて:"))
    break
  except ValueError:
    print("数字を入れてください")
print(f"{age}歳ですね")
```

```python
while True:
  try:
    food=["ハンバーガー","ポテト","ナゲット"]
    n=int(input("何番目を表示する？"))
    food[n]
    break
  except ValueError:
    print("数字を入れて")
  except IndexError:
    print("0から2までの数字を入れて")
print(f"{food[n]}ですね")
```

```python
try:
  age = int(input("年齢を入れて:"))
except ValueError:
  print("数字で入れてください")
else:
  print(f"{age}歳ですね")
finally:
  print("処理を終わります ")
```

```python
try:
  x=int(input("数字を入れて"))
  ans=10/x
except ValueError:
  print("数字じゃない")
except ZeroDivisionError:
  print("0はだめ")
else:
  print(f"答えは{ans}です")
finally:
  print("おわり")
```
